```python
"""
Dataset Loading Script
==========================================
This script loads 15 benchmark datasets used in the comparative study.
Each dataset is loaded from its original source (UCI, OpenML, or scikit-learn).

Requirements:
  pip install openml scikit-learn pandas numpy ucimlrepo
"""
!pip install openml scikit-learn pandas numpy ucimlrepo
import numpy as np
import pandas as pd
import openml
from sklearn.datasets import fetch_openml, fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from ucimlrepo import fetch_ucirepo
import urllib.request
import warnings
warnings.filterwarnings('ignore')
```

```python
# ============================================================================
# D1: Pima Indians Diabetes (OpenML ID 42608)
# ============================================================================
print("Loading D1: Pima Indians Diabetes...")
dataset_d1 = openml.datasets.get_dataset(42608)
df_d1 = dataset_d1.get_data(dataset_format="dataframe")[0]
X_d1 = df_d1.drop(columns=["Outcome"])
y_d1 = df_d1["Outcome"]
print(f"  Samples: {X_d1.shape[0]}, Features: {X_d1.shape[1]}")
```

```python
# ============================================================================
# D2: Heart Disease (Cleveland) (UCI)
# ============================================================================
print("\nLoading D2: Heart Disease (Cleveland)...")
url_d2 = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"
cols_d2 = ["age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
           "thalach", "exang", "oldpeak", "slope", "ca", "thal", "target"]
df_d2 = pd.read_csv(url_d2, header=None, names=cols_d2)
df_d2 = df_d2.replace("?", pd.NA)

X_d2 = df_d2.drop(columns=["target"])
y_d2 = df_d2["target"]

mask = X_d2.notna().all(axis=1)
X_d2 = X_d2[mask].astype(float)
y_d2 = y_d2[mask].astype(int)

print(f"  Samples: {X_d2.shape[0]}, Features: {X_d2.shape[1]}")
```

```python
# ============================================================================
# D3: Adult Income (UCI via scikit-learn)
# ============================================================================
print("\nLoading D3: Adult Income...")
adult_d3 = fetch_openml(name="adult", version=2, as_frame=True, parser='auto')
df_d3 = adult_d3.frame
X_d3 = df_d3.drop(columns=["class"])
y_d3 = df_d3["class"]
print(f"  Samples: {X_d3.shape[0]}, Features: {X_d3.shape[1]}")
```

```python
# ============================================================================
# D4: Parkinson's Disease (UCI)
# ============================================================================
print("\nLoading D4: Parkinson's Disease...")
parkinsons_d4 = fetch_ucirepo(name="parkinsons")
X_d4 = parkinsons_d4.data.features.copy()
y_d4 = parkinsons_d4.data.targets.copy()
# Remove non-feature columns if present
for col in ["name", "id", "ID"]:
    if col in X_d4.columns:
        X_d4 = X_d4.drop(columns=[col])
X_d4 = X_d4.select_dtypes(include=[np.number])
print(f"  Samples: {X_d4.shape[0]}, Features: {X_d4.shape[1]}")
```

```python
# ============================================================================
# D5: Credit Card Default (OpenML ID 46374)
# ============================================================================
print("\nLoading D5: Credit Card Default...")
dataset_d5 = openml.datasets.get_dataset(46374)
X_d5, y_d5, _, _ = dataset_d5.get_data(
    dataset_format="dataframe",
    target=dataset_d5.default_target_attribute
)
print(f"  Samples: {X_d5.shape[0]}, Features: {X_d5.shape[1]}")
```

```python
# ============================================================================
# D6: In-Vehicle Coupon Recommendation (UCI ID 603)
# ============================================================================
print("\nLoading D6: In-Vehicle Coupon Recommendation...")
coupon_d6 = fetch_ucirepo(id=603)
X_d6 = coupon_d6.data.features
y_d6 = coupon_d6.data.targets
print(f"  Samples: {X_d6.shape[0]}, Features: {X_d6.shape[1]}")
```

```python
# ============================================================================
# D7: Breast Cancer Wisconsin (Diagnostic) (UCI via scikit-learn)
# ============================================================================
print("\nLoading D7: Breast Cancer Wisconsin (Diagnostic)...")
from sklearn.datasets import load_breast_cancer
data_d7 = load_breast_cancer()
X_d7 = pd.DataFrame(data_d7.data, columns=data_d7.feature_names)
y_d7 = pd.Series(data_d7.target)
print(f"  Samples: {X_d7.shape[0]}, Features: {X_d7.shape[1]}")
```

```python
# ============================================================================
# D8: QSAR Biodegradation (UCI)
# ============================================================================
print("\nLoading D8: QSAR Biodegradation...")
url_d8 = "https://archive.ics.uci.edu/ml/machine-learning-databases/00254/biodeg.csv"
urllib.request.urlretrieve(url_d8, "biodeg.csv")
data_d8 = pd.read_csv("biodeg.csv", sep=";")
X_d8 = data_d8.iloc[:, :-1]
y_d8 = data_d8.iloc[:, -1]
print(f"  Samples: {X_d8.shape[0]}, Features: {X_d8.shape[1]}")
```

```python
# ============================================================================
# D9: Spambase (UCI via fetch_ucirepo)
# ============================================================================
print("\nLoading D9: Spambase...")
spambase_d9 = fetch_ucirepo(id=94)
X_d9 = spambase_d9.data.features
y_d9 = spambase_d9.data.targets.iloc[:, 0]
print(f"  Samples: {X_d9.shape[0]}, Features: {X_d9.shape[1]}")
```

```python
# ============================================================================
# D10: MADELON (UCI) - Combined train and validation sets
# ============================================================================
print("\nLoading D10: MADELON...")
# Download and extract
url_d10 = "https://archive.ics.uci.edu/static/public/171/madelon.zip"
urllib.request.urlretrieve(url_d10, "madelon.zip")
import zipfile
with zipfile.ZipFile("madelon.zip", "r") as z:
    z.extractall("madelon")

# Load train and validation data
X_train_d10 = pd.read_csv("madelon/MADELON/madelon_train.data", sep=" ", header=None)
X_valid_d10 = pd.read_csv("madelon/MADELON/madelon_valid.data", sep=" ", header=None)
y_train_d10 = pd.read_csv("madelon/MADELON/madelon_train.labels", header=None).iloc[:, 0]
y_valid_d10 = pd.read_csv("madelon/madelon_valid.labels", header=None).iloc[:, 0]

# Remove trailing empty column and combine
X_train_d10 = X_train_d10.dropna(axis=1, how="all")
X_valid_d10 = X_valid_d10.dropna(axis=1, how="all")
X_d10 = pd.concat([X_train_d10, X_valid_d10], axis=0, ignore_index=True)
y_d10 = pd.concat([y_train_d10, y_valid_d10], axis=0, ignore_index=True)
print(f"  Samples: {X_d10.shape[0]}, Features: {X_d10.shape[1]}")
```

```python
# ============================================================================
# D11: ISOLET Classes 1 vs 2 (OpenML ID 41966)
# ============================================================================
print("\nLoading D11: ISOLET (Classes 1 vs 2)...")
dataset_d11 = openml.datasets.get_dataset(41966)
X_d11, y_d11, _, _ = dataset_d11.get_data(
    dataset_format="dataframe",
    target=dataset_d11.default_target_attribute
)
print(f"  Samples: {X_d11.shape[0]}, Features: {X_d11.shape[1]}")
```

```python
# ============================================================================
# D12: Colon Cancer (OpenML ID 1432)
# ============================================================================
print("\nLoading D12: Colon Cancer...")
dataset_d12 = openml.datasets.get_dataset(1432)
X_d12, y_d12, _, _ = dataset_d12.get_data(
    dataset_format="array",
    target=dataset_d12.default_target_attribute
)
# Convert sparse matrix to dense if needed
if hasattr(X_d12, 'toarray'):
    X_d12 = X_d12.toarray()
X_d12 = pd.DataFrame(X_d12)
y_d12 = pd.Series(y_d12)
print(f"  Samples: {X_d12.shape[0]}, Features: {X_d12.shape[1]}")
```

```python
# ============================================================================
# D13: GISETTE (UCI) - Combined train and validation sets
# ============================================================================
print("\nLoading D13: GISETTE...")
# Download and extract
url_d13 = "https://archive.ics.uci.edu/static/public/170/gisette.zip"
urllib.request.urlretrieve(url_d13, "gisette.zip")
with zipfile.ZipFile("gisette.zip", "r") as z:
    z.extractall(".")

# Load train and validation data
X_train_d13 = pd.read_csv("GISETTE/gisette_train.data", sep=" ", header=None)
X_valid_d13 = pd.read_csv("GISETTE/gisette_valid.data", sep=" ", header=None)
y_train_d13 = pd.read_csv("GISETTE/gisette_train.labels", header=None)
y_valid_d13 = pd.read_csv("gisette_valid.labels", header=None)

# Remove trailing empty column and combine
X_train_d13 = X_train_d13.iloc[:, :-1]
X_valid_d13 = X_valid_d13.iloc[:, :-1]
X_d13 = pd.concat([X_train_d13, X_valid_d13], axis=0, ignore_index=True)
y_d13 = pd.concat([y_train_d13, y_valid_d13], axis=0).iloc[:, 0]
y_d13 = y_d13.reset_index(drop=True)
print(f"  Samples: {X_d13.shape[0]}, Features: {X_d13.shape[1]}")
```

```python
# ============================================================================
# D14: 20 Newsgroups with TF-IDF (8000 features)
# ============================================================================
print("\nLoading D14: 20 Newsgroups (TF-IDF)...")
data_d14 = fetch_20newsgroups(subset="all", remove=("headers", "footers", "quotes"))
vectorizer_d14 = TfidfVectorizer(
    max_features=8000,
    min_df=5,
    stop_words="english"
)
X_d14_sparse = vectorizer_d14.fit_transform(data_d14.data)
y_d14 = data_d14.target
# Keep as sparse matrix for efficiency, convert when needed
print(f"  Samples: {X_d14_sparse.shape[0]}, Features: {X_d14_sparse.shape[1]}")
```

```python
# ============================================================================
# D15: ARCENE (UCI) - Combined train and validation sets
# ============================================================================
print("\nLoading D15: ARCENE...")
# Download and extract
url_d15 = "https://archive.ics.uci.edu/static/public/167/arcene.zip"
urllib.request.urlretrieve(url_d15, "arcene.zip")
with zipfile.ZipFile("arcene.zip", "r") as z:
    z.extractall("arcene")

# Load train and validation data
X_train_d15 = pd.read_csv("arcene/ARCENE/arcene_train.data", sep=r"\s+", header=None)
X_valid_d15 = pd.read_csv("arcene/ARCENE/arcene_valid.data", sep=r"\s+", header=None)
y_train_d15 = pd.read_csv("arcene/ARCENE/arcene_train.labels", header=None).iloc[:, 0]
y_valid_d15 = pd.read_csv("arcene/arcene_valid.labels", header=None).iloc[:, 0]

# Combine train and validation
X_d15 = pd.concat([X_train_d15, X_valid_d15], axis=0, ignore_index=True)
y_d15 = pd.concat([y_train_d15, y_valid_d15], axis=0, ignore_index=True)
print(f"  Samples: {X_d15.shape[0]}, Features: {X_d15.shape[1]}")
```
